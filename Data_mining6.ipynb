```python
!pip install mlxtend -q

import pandas as pd
import time
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules

transactions = [
    ['Milk', 'Bread', 'Butter'],
    ['Milk', 'Bread', 'Diaper', 'Beer'],
    ['Milk', 'Bread', 'Diaper', 'Beer'],
    ['Bread', 'Butter'],
    ['Milk', 'Diaper', 'Beer'],
    ['Bread', 'Diaper', 'Beer'],
    ['Milk', 'Bread', 'Butter'],
    ['Milk', 'Bread', 'Diaper', 'Beer'],
    ['Bread', 'Butter', 'Cheese'],
    ['Milk', 'Diaper', 'Beer']
]

te = TransactionEncoder()
df = pd.DataFrame(
    te.fit_transform(transactions),
    columns=te.columns_
)

print(df.astype(int))
```

Each row represents a transaction. 1 means the item is present and 0 means it is absent.

##ii) Minimum support 0.3

```python
freq03 = apriori(df, min_support=0.3, use_colnames=True)

print(freq03.to_string(index=False))
print("Number of frequent itemsets:", len(freq03))
```

Result: 17 frequent itemsets

##iii)-iv) Association rules with minimum confidence =0.7

```python
rules03 = association_rules(
    freq03, metric='confidence', min_threshold=0.7
)

print(rules03[
    ['antecedents', 'consequents',
     'support', 'confidence', 'lift']
].round(3).to_string(index=False))

print("Number of rules:", len(rules03))
```

Result: 23 association rules.

##6.b i)

```python
start = time.perf_counter()
ap = apriori(df, min_support=0.2, use_colnames=True)
ap_time = time.perf_counter() - start

print(ap.to_string(index=False))
print("Frequent itemsets:", len(ap))
print("Execution time:", ap_time)
```

Results : 19 frequent itemsets

##b (ii)

```python
start = time.perf_counter()
fp = fpgrowth(df, min_support=0.2, use_colnames=True)
fp_time = time.perf_counter() - start

print(fp.to_string(index=False))
print("Frequent itemsets:", len(fp))
print("Execution time:", fp_time)
```

Results : 19 frequent itemsets

##b iii)

```python
rules_ap = association_rules(
    ap, metric='lift', min_threshold=1.0
)

rules_fp = association_rules(
    fp, metric='lift', min_threshold=1.0
)

cols = ['antecedents', 'consequents',
        'support', 'confidence', 'lift']

print("Apriori rules:")
print(rules_ap[cols].round(3).to_string(index=False))

print("\nFP-Growth rules:")
print(rules_fp[cols].round(3).to_string(index=False))
```

Both algorithms generate the rules whose lift is at least 1.0. The code displays support, confidence, and lift for each rule.

##iv)

```python
print("Apriori time:", ap_time, "seconds")
print("FP-Growth time:", fp_time, "seconds")
```

Fp grfowth takes less time to exicute than apriori.

##b v)

```python
def convert_itemsets(result):
    return {
        tuple(sorted(itemset)): round(support, 10)
        for itemset, support in zip(
            result['itemsets'], result['support']
        )
    }

print("Same frequent itemsets:",
      convert_itemsets(ap) == convert_itemsets(fp))
```

Result: True. Both algorithms produce the same frequent itemsets and support values.
