```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import adjusted_rand_score
from scipy.cluster.hierarchy import dendrogram,linkage

```

```python
#Generate 2D dataset
X,y_true=make_blobs(n_samples=100,
                    centers=3,
                    n_features=2,
                    random_state=42)
```

```python
#i)Agglomerative Hierarchical Clustering Using Ward Linkage
Z=linkage(X,method="ward")
print(Z)

```

```python
#ii)construct dendrogram
plt.figure(figsize=(12,5))
dendrogram(Z)
plt.title("Dendrogram")
plt.xlabel("Samples")
plt.ylabel("Euclidean Distance")
plt.show()
```

```python
#iv)
model = AgglomerativeClustering(n_clusters=3, linkage='ward')
y_pred = model.fit_predict(X)
print(y_pred)
```

```python
# Plot identified clusters
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap='viridis')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('Agglomerative Hierarchical Clustering')
plt.show()
```

```python
# 5. Calculate ARI
ari = adjusted_rand_score(y_true, y_pred)
print("Adjusted Rand Index (ARI):", round(ari, 4))
```

Interpretation:
Number of clusters: The dendrogram shows how observations merge into larger clusters. A large vertical gap between successive merges can suggest a suitable place to cut the dendrogram. Here, we use 3 clusters, consistent with the dataset's three generating centers; this should be checked against the plotted dendrogram.
ARI interpretation:
ARI = 1: Perfect agreement with the true cluster labels.
ARI ≈ 0: Agreement similar to random clustering.
ARI < 0: Agreement worse than expected by chance.
Conclusion: A higher ARI indicates that the hierarchical clustering solution more closely matches the original cluster structure. The actual ARI value is printed by the code.
